```python
import numpy as np
import pandas as pd
from datetime import datetime, timedelta
from alpaca.data.historical import CryptoHistoricalDataClient
from alpaca.data.requests import CryptoBarsRequest
from alpaca.data.timeframe import TimeFrame

client = CryptoHistoricalDataClient()
client
```

# Get data

```python
request_params = CryptoBarsRequest(
  symbol_or_symbols=["BTC/USD"],
  timeframe=TimeFrame.Hour,
  start=datetime(2024, 1, 1),
  end=datetime(2025, 12, 31)
)

btc_bars = client.get_crypto_bars(request_params)
btc_bars = btc_bars.df

btc_bars
```

```python
btc_bars['Result'] = np.where(btc_bars['close'] > btc_bars['open'], 1, 0)
btc_bars['Result'].mean()
```

# Transform data

```python
for i in range(5):
    btc_bars[f"hour_{i+1}"] = btc_bars['Result'].shift(i+1).astype('Int64')

btc_bars = btc_bars.dropna()
btc_bars.head()
```

```python
import pandas as pd
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report

# 1. Prepare your data
btc_bars['date'] = pd.to_datetime(btc_bars.index.get_level_values(1))
btc_bars = btc_bars.sample(frac=1)

# Define features (X) and target (y)
X = btc_bars[['open', 'high', 'low', 'close', 'hour_1', 'hour_2', 'hour_3', 'hour_4', 'hour_5']]
y = btc_bars['Result']

# 2. Split data by date
# Example: 70% Train, 15% Validation, 15% Test
n = len(btc_bars)
train_end = int(n * 0.70)
val_end = int(n * 0.85)

X_train, y_train = X.iloc[:train_end], y.iloc[:train_end]
X_val, y_val = X.iloc[train_end:val_end], y.iloc[train_end:val_end]
X_test, y_test = X.iloc[val_end:], y.iloc[val_end:]

# 3. Build and Train the Random Forest Classifier
clf = RandomForestClassifier(n_estimators=100, random_state=54)
clf.fit(X_train[['hour_1', 'hour_2', 'hour_3', 'hour_4', 'hour_5']], y_train)

# 4. Evaluate the model
val_preds = clf.predict(X_val[['hour_1', 'hour_2', 'hour_3', 'hour_4', 'hour_5']])
print("Validation Accuracy:", accuracy_score(y_val, val_preds))

# Performance
test_preds = clf.predict(X_test[['hour_1', 'hour_2', 'hour_3', 'hour_4', 'hour_5']])
print("Test Accuracy:", accuracy_score(y_test, test_preds))
print(classification_report(y_test, test_preds))
```

```python
results = pd.DataFrame({'Open': X_test['open'], 'Close': X_test['close'], 'Actual': y_test, 'Predicted': test_preds})
results['Result'] = np.where(results['Actual'] == results['Predicted'], 1, 0)
results['difference'] = (results['Close'] - results['Open']) / results['Open']

results['Year'] = results.index.get_level_values(1).year
results['Month'] = results.index.get_level_values(1).month
results
```

```python
# Average accuracy
results['Result'].mean(), results['Actual'].mean(), results['Predicted'].mean()
```

```python
# Calculate per year and month the number of rows and the accuracy
results.groupby('Year').agg(
    Trades=('Result', 'size'),
    Positive_trades=('Predicted', 'sum'),
    Accuracy=('Result', 'mean')
).reset_index()
```

```python
# Calculate the total percentage gain based on the predictions and add print it as a percentage
str(round((results[results['Predicted'] == 1].difference.sum() - results[results['Predicted'] == 0].difference.sum())*100, 1)) + '%'
```
